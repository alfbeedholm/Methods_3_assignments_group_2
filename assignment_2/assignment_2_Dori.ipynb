&nbsp;&nbsp;&nbsp;&nbsp;    vii. Create a new reduced $X$ that only includes the non-zero coefficients - show the covariance of the non-zero features; $X_{reduced}X_{reduced}^T$ . Plot the covariance of the features using `plt.imshow`. Compared to a covariance-plot where you do not use any penalties, do you see less covariance? (Ignore the absolute values, but look at the pattern)


```python
# log_l1.coef_ is a 2D array, with 1 row of coefficient indexes; we take this row and make it into a 1D arrray with [0]
# nonz_index = np.nonzero(log_l1.coef_[0])
# nonz_index is a touple, with the list of feature indexes as the first item
nonz_index = np.nonzero(log_l1.coef_[0])[0] # taking only the list of feature indexes

X_reduced = X_1_2_s[:, nonz_index].T # because it has trials as rows and features as columns, we transpose first
# if we do not transpose, we would get dimensions cov = (trials × features) * (features × trials) =  (trials × trials) when we want (features × featues)
cov_nonz = X_reduced @ (X_reduced).T
plt.imshow(cov_nonz)
plt.colorbar()
plt.title('Penalized Logistic Regression Covariance Matrix')
```

```python
# comparing it with the covariance matrix of the non-penalized model
X_full = X_1_2_s[:, ::10].T # taking every 10th feature, because otherwise the plot would need to display 25602 * 25602 pixels.
plt.imshow(X_full @ X_full.T)
plt.colorbar()
plt.title(f'No Penalty: Every 10th of {X_1_2_s.shape[1]} Features')
```

There is much less covariance between the features of the penalised model. With the penalised model, we see covariance almost only along the diagonal, which shows that most features don't move together. Whereas, with the non-penalised features' covariance matrix, we see clear covariance between many features; this means that the number of these features could be further reduced to avoid unnecessary model complexity.

```python
# Exercise B.2.ii
def equalize_targets_binary(data, y):
    np.random.seed(7)
    targets = np.unique(y) ## find the number of targets
    if len(targets) > 2:
        raise NameError("can't have more than two targets")
    counts = list()
    indices = list()
    for target in targets:
        counts.append(np.sum(y == target)) ## find the number of each target
        indices.append(np.where(y == target)[0]) ## find their indices
    min_count = np.min(counts)
    # randomly choose trials
    first_choice = np.random.choice(indices[0], size=min_count, replace=False)
    second_choice = np.random.choice(indices[1], size=min_count,replace=False)
    
    # create the new data sets
    new_indices = np.concatenate((first_choice, second_choice))
    new_y = y[new_indices]
    new_data = data[new_indices, :, :]
    
    return new_data, new_y
```
